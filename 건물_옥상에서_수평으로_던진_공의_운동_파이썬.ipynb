```python
import math
```

```python
# 사용자 입력

t = float(input("공이 떨어지기까지 걸린 시간 (초)을 입력하세요 (예: 2): "))

x = float(input("수평 거리 (m)를 입력하세요 (예: 20): "))
```

```python
g = 9.81  # m/s^2
```

```python
# a) 공의 처음속도 (수평속도)

v0 = x / t
```

```python
# b) 옥상의 높이 (자유낙하 공식 사용)

y = 0.5 * g * t**2
```

```python
# c) 공이 지면에 닿을 때 속도

vx = v0

vy = g * t
```

```python
# 최종 속도 크기

v = math.sqrt(v0**2 + (g * t)**2)
```

```python
# 최종 속도 방향 (아래로 향한 각도, 단위: 도)

theta_rad = math.atan(vy / vx)  # atan() 앞에 math. 붙이기
theta_deg = math.degrees(theta_rad)  # 라디안을 도(degree)로 변환
```

```python
#결과 출력

print(f"\n[결과]")

print(f"a) 공의 처음속도 크기: {v0:.2f} m/s")

print(f"b) 건물 높이: {y:.2f} m")

print(f"c) 충돌 순간 속도 크기: {v:.2f} m/s")

print(f"   충돌 순간 속도 방향 (수평선과 이루는 각도): {theta_deg:.2f}도")
```
