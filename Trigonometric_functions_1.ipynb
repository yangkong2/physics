# 직교 좌표를 극좌표로 변환

```python
x, y = 1.0, 2.0
```

```python
r = math.sqrt(x**2 + y**2)
theta_raidan = math.atan2(y, x)

print(r, theta_raidan)
```

# 극좌표를 직교 좌표로 변환

```python
r= 10
angle = 30
```

```python
x = r * math.cos(angle)
y = r * math.sin(angle)

print(x ,y)
```
